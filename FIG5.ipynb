```python
import numpy as np
import h5py
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
from cartopy.util import add_cyclic_point
from matplotlib import cm
plt.rcParams['font.family'] = 'Arial'
plt.rcParams['svg.fonttype']='none'
plt.rcParams['mathtext.fontset'] = 'custom'
plt.rcParams['mathtext.rm'] = 'Arial'
plt.rcParams['mathtext.it'] = 'Arial:italic'
plt.rcParams['mathtext.bf'] = 'Arial:bold'
```

```python
dis_address = '../data/Same_ET_ST_result/distance.h5'
f_save = h5py.File(dis_address, 'r')
distance_array = f_save['distance'][:]
lat = f_save['lat'][:]
lon = f_save['lon'][:]
f_save.close()
```

```python
def compute (SC,SC_ij,SC_ji):

    degree_5000 = np.zeros((73,144))
    for m in range (0,73,1):
        for n in range (0,144,1):

            degree_value = 0

            for p in range (0,73,1):
                for q in range (0,144,1):
                    if SC[m,n,p,q] == 0:
                        continue
                    else:
                        if distance_array[m,n,p,q]>5000:
                            degree_value += abs(SC[m,n,p,q]*np.cos(lat[p]*np.pi/180))
                        else:
                            continue
            degree_5000[m,n] = degree_value

    in_degree = np.zeros((73,144))
    out_degree = np.zeros((73,144))

    for m in range (0,73,1):
        for n in range (0,144,1):

            in_degree_value = 0
            out_degree_value = 0

            for p in range (0,73,1):
                for q in range (0,144,1):
                    if SC_ij[m,n,p,q] == 0:
                        continue
                    else:
                        if distance_array[m,n,p,q]<=4000:
                            in_degree_value += SC_ji[m,n,p,q]*np.cos(lat[p]*np.pi/180)
                            out_degree_value += SC_ij[m,n,p,q]*np.cos(lat[p]*np.pi/180)

                        else:
                            continue
            in_degree[m,n] = in_degree_value
            out_degree[m,n] = out_degree_value

    divergence = out_degree-in_degree

    return degree_5000,divergence
```

```python
Network2_address = '../data/Same_ET_ST_result/Network2_SC.h5'
f_Network2 = h5py.File(Network2_address,'r')
Network2_SC = f_Network2['SC'][:]
lat = f_Network2['lat'][:] 
lon = f_Network2['lon'][:]
f_Network2.close()

Network2_address = '../data/Same_ET_ST_result/Network2_SC_in_out.h5'
f_Network2 = h5py.File(Network2_address,'r')
Network2_SC_ij = f_Network2['SC_ij'][:]
Network2_SC_ji = f_Network2['SC_ji'][:]
f_Network2.close()

p1_degree_5K,P1_divergence = compute(Network2_SC,Network2_SC_ij,Network2_SC_ji)
```

```python
def bigger300_W_north(array):

    temp_north_degree = array[36:,]
    temp_north_degree = np.reshape(temp_north_degree,-1)
    temp_W = temp_north_degree[temp_north_degree>300]

    return np.array(temp_W)
```

```python
def bigger300_W_south(array):
    temp_north_degree = array[:37,]
    temp_north_degree = np.reshape(temp_north_degree,-1)
    temp_W = temp_north_degree[temp_north_degree>300]
    return np.array(temp_W)
```

```python
def bigger5_D_north(array):

    temp_north_degree = array[36:,]
    temp_north_degree = np.reshape(temp_north_degree,-1)
    temp_north_degree = abs(temp_north_degree)
    temp_D = temp_north_degree[temp_north_degree>5]

    return np.array(temp_D)
```

```python
def bigger5_D_south(array):

    temp_south_degree = array[:37,]
    temp_south_degree = np.reshape(temp_south_degree,-1)
    temp_south_degree = abs(temp_south_degree)
    temp_D = temp_south_degree[temp_south_degree>5]

    return np.array(temp_D)
```
